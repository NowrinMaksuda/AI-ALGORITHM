**Dataset:** Pima Indians Diabetes Dataset — 768 patients, 8 medical features (`Pregnancies`, `Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, `BMI`, `DiabetesPedigreeFunction`, `Age`) and a binary target `Outcome` (1 = diabetic, 0 = not diabetic).

**Goal:** Implement the KNN algorithm **completely from scratch using pure Python** — plain `list`s, `for` loops, and the built-in `math`/`random`/`csv` modules only. **No `numpy`, `pandas`, or `sklearn.neighbors` anywhere.** `matplotlib` is used only at the end to draw a results chart.

### Algorithm (as per assignment brief)
1. Look at the new point (from the testing data set) you want to predict.
2. Calculate the distance from this new point to every point in the training data.
3. Use a distance formula — Euclidean distance (straight-line distance).
4. Sort all the points by their distance to the new point (closest first).
5. Pick the top K closest points (these are the neighbors).
6. Check the labels of those K neighbors — for classification, the class with the most votes among the neighbors wins.
7. Repeat for every point in the testing data set, then calculate the accuracy of the model.

Run for **K = 1, 3, 5, 7** and report accuracy for each.

## 2. Build Feature Vectors and Labels (plain Python lists)

Each patient becomes a plain Python `list` of 8 numbers (the features), plus a separate label (0 or 1). No `numpy` arrays, no `pandas` DataFrame — just lists of lists.


```python
feature_names = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
                  'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']

raw_data = []
labels = []
for row in rows:
    point = [float(row[feat]) for feat in feature_names]
    raw_data.append(point)
    labels.append(int(float(row['Outcome'])))

print("Total samples:", len(raw_data))
print("Example features:", raw_data[0])
print("Example label:", labels[0])
print("Class distribution -> Outcome=1 (diabetic):", labels.count(1), " Outcome=0 (not diabetic):", labels.count(0))
```

## 3. Manual Standardization (z-score, no `numpy`/`StandardScaler`)

KNN relies entirely on distance, so features must be on a comparable scale — otherwise `Insulin`/`Glucose` (large values) would dominate `DiabetesPedigreeFunction` (small decimal values).

$$z = \frac{x - \mu}{\sigma}$$


```python
def column(data, index):
    return [point[index] for point in data]


def mean(values):
    return sum(values) / len(values)


def std_dev(values, mu):
    variance = sum((x - mu) ** 2 for x in values) / len(values)
    return math.sqrt(variance)


def standardize(data, n_features):
    means = []
    stds = []
    for j in range(n_features):
        col = column(data, j)
        mu = mean(col)
        sigma = std_dev(col, mu)
        means.append(mu)
        stds.append(sigma if sigma > 0 else 1.0)  # guard against zero-variance columns

    standardized = []
    for point in data:
        standardized.append([(point[j] - means[j]) / stds[j] for j in range(n_features)])

    return standardized, means, stds


data, feat_means, feat_stds = standardize(raw_data, len(feature_names))

print("Feature means:", [round(m, 3) for m in feat_means])
print("Feature stds :", [round(s, 3) for s in feat_stds])
```

## 4. Manual Train/Test Split (no `sklearn.model_selection`)

We shuffle the data with a fixed random seed (for reproducibility) and split it — 80% train, 20% test — using plain Python.


```python
def train_test_split_manual(data, labels, test_ratio=0.2, seed=42):
    n = len(data)
    indices = list(range(n))
    rng = random.Random(seed)
    rng.shuffle(indices)

    n_test = int(n * test_ratio)
    test_idx = indices[:n_test]
    train_idx = indices[n_test:]

    X_train = [data[i] for i in train_idx]
    y_train = [labels[i] for i in train_idx]
    X_test = [data[i] for i in test_idx]
    y_test = [labels[i] for i in test_idx]

    return X_train, y_train, X_test, y_test


X_train, y_train, X_test, y_test = train_test_split_manual(data, labels, test_ratio=0.2, seed=42)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

## 5. Manual KNN Implementation (pure Python)

Every step of the assignment's algorithm — Euclidean distance, sorting by distance, picking the K nearest neighbors, and majority-vote classification — is written with plain Python `list`s and `for` loops.


```python
def euclidean_distance(point_a, point_b):
    """Step 3: Euclidean (straight-line) distance between two points."""
    total = 0.0
    for a, b in zip(point_a, point_b):
        total += (a - b) ** 2
    return math.sqrt(total)


def get_neighbors(X_train, y_train, test_point, k):
    """Steps 2, 3, 4, 5: compute distance to every training point,
    sort by distance, and return the labels of the k closest ones."""
    distances = []
    for i in range(len(X_train)):
        d = euclidean_distance(test_point, X_train[i])
        distances.append((d, y_train[i]))

    # Step 4: sort all points by distance (closest first) -- manual sort, no numpy
    distances.sort(key=lambda pair: pair[0])

    # Step 5: pick the top k closest points
    neighbors = distances[:k]
    return neighbors


def majority_vote(neighbors):
    """Step 6: the class with the most votes among the k neighbors wins."""
    votes = {}
    for _, label in neighbors:
        votes[label] = votes.get(label, 0) + 1

    best_label = None
    best_count = -1
    for label, count in votes.items():
        if count > best_count:
            best_count = count
            best_label = label
    return best_label


def knn_predict(X_train, y_train, test_point, k):
    neighbors = get_neighbors(X_train, y_train, test_point, k)
    return majority_vote(neighbors)


def knn_predict_all(X_train, y_train, X_test, k):
    """Step 7: repeat prediction for every point in the testing set."""
    predictions = []
    for test_point in X_test:
        pred = knn_predict(X_train, y_train, test_point, k)
        predictions.append(pred)
    return predictions
```

## 6. Manual Accuracy Metric

$$\text{Accuracy} = \frac{\text{number of correct predictions}}{\text{total number of predictions}}$$


```python
def accuracy_score_manual(y_true, y_pred):
    correct = sum(1 for actual, predicted in zip(y_true, y_pred) if actual == predicted)
    return correct / len(y_true)
```

## 7. Run KNN for K = 1, 3, 5, 7 and Report Accuracy


```python
k_values = [1, 3, 5, 7]
results = {}

for k in k_values:
    predictions = knn_predict_all(X_train, y_train, X_test, k)
    acc = accuracy_score_manual(y_test, predictions)
    results[k] = {'predictions': predictions, 'accuracy': acc}
    print(f"K={k}: Accuracy = {acc:.4f}  ({acc*100:.2f}%)")
```

## 8. Confusion Matrix (manual) for the Best K

A simple manually-built confusion matrix — how many predictions were correct/incorrect for each class — for the K with the highest accuracy.


```python
best_k = max(results, key=lambda k: results[k]['accuracy'])
best_predictions = results[best_k]['predictions']

def confusion_matrix_manual(y_true, y_pred):
    tp = sum(1 for a, p in zip(y_true, y_pred) if a == 1 and p == 1)
    tn = sum(1 for a, p in zip(y_true, y_pred) if a == 0 and p == 0)
    fp = sum(1 for a, p in zip(y_true, y_pred) if a == 0 and p == 1)
    fn = sum(1 for a, p in zip(y_true, y_pred) if a == 1 and p == 0)
    return tp, tn, fp, fn

tp, tn, fp, fn = confusion_matrix_manual(y_test, best_predictions)

print(f"Best K = {best_k} (Accuracy = {results[best_k]['accuracy']:.4f})\n")
print("Confusion Matrix:")
print(f"                  Predicted 0   Predicted 1")
print(f"  Actual 0  |     {tn:>10}   {fp:>10}")
print(f"  Actual 1  |     {fn:>10}   {tp:>10}")

precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f"\nPrecision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")
```
